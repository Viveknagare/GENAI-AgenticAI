```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from serpapi_search_tools import maps_search, news_search, web_search
from langgraph.checkpoint.memory import InMemorySaver
```

```python
llm = ChatOpenAI(model="gpt-4o")
```

```python
agent = create_agent(
    model=llm,
    tools=[web_search(), news_search()],
    system_prompt="You are a agent which is having access to google search so for any user query try to give the latest answer also you have assess to the memory in which you can store conversations",
    checkpointer=InMemorySaver()
)
```

```python
result = agent.invoke(
    {'messages' : [{'role' : 'user', 'content' : 'what is my name'}]},
    {'configurable' : {'thread_id' : '1'}}
    )
print(result)
```
