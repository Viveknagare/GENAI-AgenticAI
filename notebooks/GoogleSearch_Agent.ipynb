```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI
from serpapi_search_tools import maps_search, news_search, web_search
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

```python
agent = create_agent(
    model=llm,
    tools=[web_search(), news_search()],
    system_prompt="You are a agent which is having access to google search so for any user query try to give the latest answer"
)
```

```python
result = agent.invoke({'messages' : [{'role' : 'user', 'content' : 'do you have any access to google search tool now?'}]})
print(result)
```

```python
print(result['messages'])
```

```python
for res in result['messages']:
    print(res)
    print("\n")
```
