```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = "What Is System Design? System design is the process of defining the "\
    "architecture, components, interfaces, and data flow of a software" \
    "system. It helps engineers build applications that are scalable," \
    "reliable, maintainable, secure, and cost-effective."
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 50, chunk_overlap = 20)
chunks = splitter.split_text(text)
print(chunks)
```

Document text Loader


```python
from langchain_community.document_loaders import TextLoader,PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
loader = PyPDFLoader("../data/College.pdf")
text = loader.load()
```

```python
loader = TextLoader("../data/system_design_concepts.txt")
texts = loader.load()
len(texts)
print(texts[0].page_content)
```

```python
pdfloader = PyPDFLoader("../data/College.pdf")
pdf_pages = pdfloader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)
```

```python
text_chunks = splitter.split_documents(texts)
len(text_chunks)
```

```python
pdf_chunks = splitter.split_documents(pdf_pages)
len(pdf_chunks)
```

```python
for chunk in pdf_chunks:   #you can see that how chunks are overlapped
    print(chunk.page_content)
    print("\n")
```
