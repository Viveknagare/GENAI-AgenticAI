###connect with openai model


```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-4o")
```

```python
res = llm.invoke("who was the winner of ipl 2025?")
print(res.content)
```

###connect with gemini model

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

```python
res_gemini = llm.invoke("who are you")
print(res_gemini.text)
```

#Static prompts

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI()
```

```python
prompts = [
    ("system", "you are a python developer"),
    ("user", "how to add two numbers")
]
res = llm.invoke(prompts)
print(res.content)
```
