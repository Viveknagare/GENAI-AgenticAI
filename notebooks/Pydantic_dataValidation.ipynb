```python
from pydantic import BaseModel, Field
```

```python
class User(BaseModel):
    name : str
    age : int
```

```python
user = User(name="vivek", age=23)
```

```python
print(user)
```

#### Field Validation

```python
class Product(BaseModel):
    title : str = Field(max_length=10)
    desc : str = Field(min_length=20)
    price : int = Field(gt=0, le=100000)
```

```python
product1 = Product(title="iphone17", desc="This is an amazing phone with lots of features and stable UI", price=99990)
product2 = Product(title="Samsungs26", desc="Best Value for money phone under 80k", price=79990)
```

```python
print(product1)
```

```python
class Order(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    product:Product
```

```python
order = Order(id="7484j4i9", product=product1)  #Either you can pass product instance or dictionary of required values
```

```python
print(order)
```

```python
print(order.product.title)
```

```python
from typing import List, Annotated
```

```python
class Cart(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items: List[Product]
    comment : str = Field(default="No Comment")
```

```python
cart = Cart(id="12jk2k", items=[product1, product2])
cart
```

```python
for item in cart.items:
    print(item.title+ " and " +item.desc) 
```

#### Annotated - Helps us to build common/custom validator

```python
commonValidator = Annotated[List[Product], Field(default=[])]  ##this field is optional to pass default value, you can also pass description/metadata as an argument in it.
```

```python
class Cart(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items: commonValidator
    comment : str = Field(default="No Comment")
```

```python
class Order(BaseModel):
    products:commonValidator
```

##### whenever we have to change some validation of data and that validation is comman for most of the classes then we create a comman/custom validator for all of them and then only the validator instead of changing it in every class

```python
ord = Order(products=[product1,product2])  ##If you dont pass anything it will create a empty List as we have mentioned that default value would be []
```

```python
ord
```

```python
for item in ord.products:
    print(item.title)
```
