```python
from dotenv import load_dotenv
from langchain_core.tools import tool
```

Define a Tool


```python
@tool
def add_numbers(a:int, b:int):
    """
    It will return the sum of two numbers
    
    Args:
    a : First Number
    b : Second Number
    """
    return a + b

@tool
def sub_numbers(a: int, b: int):
    """
    It will return the subtraction of two numbers
    Args:
    a : First Number
    b : Second Number
    """
    return a - b
```

Execute Tool by giving some Arguments

```python
add_numbers.invoke({"a" : 12, "b" : 16})
```

Create an LLM

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

Connect the LLM and Tool to create a AI Agent

```python
from langchain.agents import create_agent
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers, sub_numbers],
    system_prompt="You are a Maths tutor and always use tools for calculation"
)
```

```python
response = agent.invoke({"messages":[{"role" : "user", "content" : " what is 27 + 48"}]})
response
```
