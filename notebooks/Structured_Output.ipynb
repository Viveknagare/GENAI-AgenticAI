```python
from dotenv import load_dotenv
load_dotenv()
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o")
```

Problem is LLM extract out the details and give structured output but in string format, but we want it in dictionary/json/object format where we can extract out details from it

```python
text = "My name is Vivek nagare, mail is vivekbnagare31@gmail.com and age is 23"
res = llm.invoke(f"Extract out the name and email from the {text}")
print(res.content)
print(type(res.content))
```

```python
from pydantic import BaseModel, Field
from typing import List

class ResponseStructure(BaseModel):
    name: str = Field(description="Complete name of the user")
    email: str = Field(description="email of the user")
    age: int = Field(description="Age of the user")
```

```python
structured_Output_llm = llm.with_structured_output(ResponseStructure)
res = structured_Output_llm.invoke(f"Extract out the name and email from the {text}")
res #this is the object of ResponseStructure class that we got from LLM, we can asses the values using the object
print(res.name)
print(res.email)
print(res.age)
```

```python
dict = res.model_dump()  # we can convert the response object in dictionary as well
print(dict["name"])
print(dict["email"])
print(dict["age"])
```

```python
class Movies(BaseModel):
    title:str = Field(description="Name of the movie")
    year:int = Field(description="Year in which the movie was released")
    
class Allmovies(BaseModel):
    movies: List[Movies]
    
movie_llm = llm.with_structured_output(Allmovies)
```

```python
res = movie_llm.invoke("Give me the latest 4 Bollywood movie")
```

```python
for i in res.movies:
    print(i)
```

```python
for movie in res.movies:
    print(movie.title)
```
